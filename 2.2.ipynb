```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'Day': ['Monday', 'Tuesday', 'Wednesday', 'Thursday','Friday', 'Saturday', 'Sunday'],
    'Temperature': [30, 32, np.nan, 34, np.nan, 36, 38]
})
print(df)

df_mean = df.copy()
df_mean['Temperature'] = df_mean['Temperature'].fillna(
    df_mean['Temperature'].mean()
)

print("\nMissing values filled using Mean:")
print(df_mean)

df_interpolate = df.copy()
df_interpolate['Temperature'] = df_interpolate['Temperature'].interpolate()

print("\nMissing values filled using Interpolation:")
print(df_interpolate)
```

```python
import numpy as np
import pandas as pd

index = pd.MultiIndex.from_product(
    [[2013, 2014], [1, 2]],
    names=['year', 'visit']
)
columns = pd.MultiIndex.from_product(
    [['Bob', 'Guido', 'Sue'], ['HR', 'Temp']],
    names=['name', 'type']
)
data = np.round(np.random.randn(4, 6), 1)
health_data = pd.DataFrame(data, index=index, columns=columns)

print("Health Data:")
print(health_data)
print("\nGuido HR:")
print(health_data[('Guido', 'HR')])
print("\nFirst two rows and columns:")
print(health_data.iloc[:2, :2])
print("\nBob HR:")
print(health_data.loc[:, ('Bob', 'HR')])

idx = pd.IndexSlice
print("\nHR values for visit 1:")
print(health_data.loc[idx[:, 1], idx[:, 'HR']])

index = pd.MultiIndex.from_product(
    [['a', 'c', 'b'], [1, 2]],
    names=['char', 'int']
)

data = pd.Series(np.random.rand(6), index=index)

print("\nOriginal Series:")
print(data)

data = data.sort_index()
print("\nSorted Series:")
print(data)

print("\nSlice from a to b:")
print(data['a':'b'])

pop = pd.Series(
    [100, 120, 150, 180, 200, 220],
    index=pd.MultiIndex.from_product(
        [['Karnataka', 'Kerala', 'Tamil Nadu'], [2023, 2024]],
        names=['state', 'year']
    ),
    name='population'
)

print("\nPopulation Series:")
print(pop)

print("\nUnstack level 0:")
print(pop.unstack(level=0))

print("\nUnstack level 1:")
print(pop.unstack(level=1))

print("\nUnstack and stack:")
print(pop.unstack().stack())

pop_flat = pop.reset_index(name='population')
print("\nFlattened population DataFrame:")
print(pop_flat)

print("\nSet state and year as index:")
print(pop_flat.set_index(['state', 'year']))

# 12. Calculate mean by year
print("\nMean by year:")
data_mean = health_data.groupby(level='year').mean()
print(data_mean)

# 13. Calculate mean by measurement type (HR and Temp)
print("\nMean by measurement type:")
data_mean_type = health_data.T.groupby(level='type').mean().T
print(data_mean_type)
```
