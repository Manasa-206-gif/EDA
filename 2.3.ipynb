```python
import pandas as pd
index = pd.MultiIndex.from_tuples([
    ('Karnataka', 2010),
    ('Karnataka', 2020),
    ('Maharashtra', 2010),
    ('Maharashtra', 2020),
    ('Tamil Nadu', 2010),
    ('Tamil Nadu', 2020)
], names=['state', 'year'])
population = pd.Series(
    [611, 675, 1124, 1240, 721, 778],
    index=index
)
print(population)
```

```python
print(population.loc['Karnataka'])
print(population.xs(2020, level='year'))
print(population.loc[('Karnataka', 2020)])
```

```python
df = population.unstack()
print("DataFrame:")
print(df)

s = df.stack()
print("Series:")
print(s)
```

```python
import pandas as pd
import numpy as np

index = pd.MultiIndex.from_product(
    [['CSE', 'AIML'], [2023, 2024]],
    names=['department', 'year']
)

df = pd.DataFrame(
    np.random.randint(50, 100, size=(4, 2)),
    index=index,
    columns=['Marks', 'Attendance']
)

print(df)
```

```python
import pandas as pd
import numpy as np

row_index = pd.MultiIndex.from_product(
    [['Python', 'DBMS'], ['Student1', 'Student2']],
    names=['subject', 'student']
)

col_index = pd.MultiIndex.from_product(
    [['Term1', 'Term2'], ['Internal', 'External']],
    names=['term', 'exam_type']
)

df = pd.DataFrame(
    np.random.randint(40, 100, size=(4, 4)),
    index=row_index,
    columns=col_index
)

print(df)
```

```python

print(df.loc['Python'])
print(df.T.groupby(level='term').mean().T)
```

```python
import pandas as pd

index = pd.MultiIndex.from_tuples([
    ('StateB', 2020),
    ('StateA', 2010),
    ('StateC', 2020),
    ('StateA', 2020),
    ('StateB', 2010)
], names=['state', 'year'])

data = pd.Series([200, 100, 300, 150, 180], index=index)

try:
    print(data.loc['StateA':'StateB'])
except Exception as e:
    print("Error:", e)
data = data.sort_index()
print("After sorting:")
print(data.loc['StateA':'StateB'])
```

```python
import pandas as pd

df = pd.DataFrame({
    'state': ['Karnataka', 'Karnataka',
              'Maharashtra', 'Maharashtra'],
    'year': [2010, 2020, 2010, 2020],
    'population': [611, 675, 1124, 1240]
})

multi_df = df.set_index(['state', 'year'])
print("MultiIndex DataFrame:")
print(multi_df)

flat_df = multi_df.reset_index()
print("Normal DataFrame:")
print(flat_df)
```
