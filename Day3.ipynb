```python
# 5
import numpy as np
arr=np.arange(1,21).reshape(4,5) # larger array
view=arr[:3,:4] # view-changes affect the original array
view[0,0]=100 
print(arr)
copy_arr=arr[:3,:4].copy() # copy array-changes do not affect the original array
copy_arr[0,0]=200
print(arr) 
print(copy_arr) 
```

```python
# 6
import numpy as np
x=np.arange(1,13)
print(x)
matrix=x.reshape(3,4) # 3*4 matrix
print(matrix)
column1=x.reshape((12,1)) # reshape column vector using reshpe(12,1)
print(column1)
column2=x[:,np.newaxis] # reshape using x[:,np.newaxis]
print(column2)
print(np.array_equal(column1,column2)) # check both  results are identical
```

```python
# 7
import numpy as np
arr1=np.random.randint(1,10,size=(2,3))
arr2=np.random.randint(1,10,size=(2,3))
print(arr1)
print(arr2)
vertical1=np.vstack((arr1,arr2)) # join arrays vertically(row-wise)
vertical2=np.concatenate((arr1,arr2),axis=0) # also join row-wise
print(vertical1)
print(vertical2)
print(np.array_equal (vertical1,vertical2)) # check both results are identical
horizontal1=np.hstack((arr1,arr2)) # join arrays horizontally(column wise)
horizontal2=np.concatenate((arr1,arr2),axis=1) # also join column-wise
print(horizontal1)
print(horizontal2)
print(np.array_equal (horizontal1,horizontal2)) # check both results are identical
```
