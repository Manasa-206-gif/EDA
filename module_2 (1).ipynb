```python
import pandas as pd
population = pd.Series([20, 19, 13, 12, 15],index=["Mumbai", "Delhi", "Bengaluru", "Chennai", "Kolkata"])
print(population)
print("\nPopulation of Bengaluru:")
print(population["Bengaluru"])
print("\nDelhi to Chennai:")
print(population["Delhi":"Chennai"])
```

```python
import pandas as pd
area = pd.Series([307713, 342239, 191791, 196024, 308252],index=["Rajasthan", "Madhya Pradesh", "Karnataka", "Gujarat", "Maharashtra"])
population = pd.Series([81, 72, 67, 64, 126],index=["Rajasthan", "Madhya Pradesh", "Karnataka", "Gujarat", "Maharashtra"])
states = pd.DataFrame({"area": area,"population": population})
states["density"] = states["population"] / states["area"]
print(states)
```

```python
import pandas as pd
indA = pd.Index([2, 3, 5, 7, 11])
indB = pd.Index([1, 3, 5, 7, 9])
print("Intersection:")
print(indA.intersection(indB))
print("\nUnion:")
print(indA.union(indB))
print("\nSymmetric Difference:")
print(indA.symmetric_difference(indB))
```
