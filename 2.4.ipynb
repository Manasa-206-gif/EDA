```python
import pandas as pd
data = {
    'department': ['CSE', 'AIML', 'ECE', 'CSE', 'AIML', 'ECE',
                   'CSE', 'AIML', 'ECE', 'CSE', 'AIML', 'ECE'],
    'gender': ['M', 'F', 'M', 'F', 'M', 'F',
               'M', 'F', 'F', 'F', 'M', 'M'],
    'salary': [40000, 45000, 35000, 42000, 50000, 38000,
               45000, 48000, 40000, 43000, 52000, 37000],
    'age': [22, 28, 35, 24, 32, 42, 26, 29, 45, 23, 38, 50]
}
df = pd.DataFrame(data)
df['age_category'] = pd.cut(
    df['age'],
    bins=[18, 30, 40, 60],
    labels=['Junior', 'Mid', 'Senior']
)
print(df)
print()
pivot = pd.pivot_table(
    df,
    values='salary',
    index='age_category',
    columns='department',
    aggfunc='mean',
    observed=False
)
print(pivot)
```

```python
import pandas as pd
data = {
    'department': ['CSE', 'AIML', 'ECE', 'CSE', 'AIML', 'ECE',
                   'CSE', 'AIML', 'ECE', 'CSE', 'AIML', 'ECE'],
    'gender': ['M', 'F', 'M', 'F', 'M', 'F',
               'M', 'F', 'F', 'F', 'M', 'M'],
    'salary': [40000, 45000, 35000, 42000, 50000, 38000,
               45000, 48000, 40000, 43000, 52000, 37000],
    'age': [22, 28, 35, 24, 32, 42, 26, 29, 45, 23, 38, 50]
}
df = pd.DataFrame(data)
pivot = pd.pivot_table(
    df,
    index='department',
    aggfunc={'salary': 'sum', 'age': 'mean'}
)

print(pivot)
```
